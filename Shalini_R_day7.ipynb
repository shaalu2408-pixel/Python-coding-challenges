# **Day 7– Data Cleaning & Stats**

# Detecting & removing duplicates.


```python
import pandas as pd
df = pd.read_csv('/content/Titanic dataset.csv')
df
```

```python
df.drop_duplicates(inplace=True)
df
```

# **Handle inconsistent or missing data.**

# **Handle Inconsistent**

```python
df.set_index('PassengerId',inplace=True)              # PassengerID as index
df
```

```python
df['Sex']=df['Sex'].str.lower().str.strip()
df['Embarked']=df['Embarked'].str.upper().str.strip()
```

```python
df['Sex']=df['Sex'].str.capitalize()                      # captilize the first word of each row in Sex column
df
```

```python
df['Last_Name'] = df['Name'].str.split(',').str[0]                              # adding a new column as Last_Name
df['Title'] = df['Name'].str.extract(r',\s*([^.]*)\.')[0].str.strip()          # adding a new column as Title
df['First_Name'] = df['Name'].str.split('.').str[-1].str.strip()               # Adding a new column as First_Name
df.drop('Name', axis=1, inplace=True)                                          # delete a column Name
df['Full_Name'] = df['First_Name'] + ' ' + df['Last_Name']                      # update a column
df
```

# **Missing Values**

```python
Embarked_mode = df['Embarked'].mode()[0]
df['Embarked'].fillna(Embarked_mode, inplace=True)
Cabin_mode = df['Cabin'].mode()[0]
df['Cabin'].fillna(Cabin_mode,inplace=True)
df['Age'].fillna(df['Age'].mean(),inplace=True)
df.info()
```

# **Compute basic statistics: mean, median, mode, std, variance.**

```python
df.median(numeric_only=True)
```

```python
df.mode(numeric_only=True)
```

```python
df.std(numeric_only=True)
```

# **Group by categories and summarize insights.**

```python
s=df.groupby('Sex').agg(
    Survival_Rate=('Survived','mean'),
    Average_age=('Age','mean'),
    Average_Fare=('Fare','mean')
     )
print(s)
```
