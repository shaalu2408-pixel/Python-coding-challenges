# **Day 2– Pandas Basics**

## **Import dataset into Pandas DataFrame.**

```python
import pandas as pd
df = pd.read_csv('/content/Titanic dataset.csv')
df
```

# **Data Cleaning**

```python
df.drop_duplicates()
```

```python
df.info()
```

```python
Embarked_mode = df['Embarked'].mode()[0]
df['Embarked'].fillna(Embarked_mode, inplace=True)
df.info()
```

```python
Cabin_mode = df['Cabin'].mode()[0]
df['Cabin'].fillna(Cabin_mode,inplace=True)
df.info()
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
df.info()
```

# **Explore series & columns.**

```python
df.columns          # explore all columns
```

```python
df['Fare']           # specific column as a Series
```

## **Add, update, delete columns/rows.**

```python
df['Last_Name'] = df['Name'].str.split(',').str[0]         # adding a new column as Last_Name
df
```

```python
df['Title'] = df['Name'].str.extract(r',\s*([^.]*)\.')[0].str.strip()          # adding a new column as Title
df
```

```python
df['First_Name'] = df['Name'].str.split('.').str[-1].str.strip()         # Addind a new column as First_Name
df
```

```python
df.drop('Name', axis=1, inplace=True)           # delete a column Name
df
```

```python
df['Full_Name'] = df['First_Name'] + ' ' + df['Last_Name']     # update a column
df
```

# **Set a custom index and sort.**

```python
df.set_index('PassengerId', inplace=True)               # PassengerID as index
df
```

```python
df.sort_values(by='Age', inplace=True)              # sorting Age column in ascending order
df
```
